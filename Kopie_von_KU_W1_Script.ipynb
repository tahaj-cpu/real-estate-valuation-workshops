# Code Script
A simple script to practice Python/Jupyter basics

```python
# Step 1: Print a greeting
print("Welcome to AI/ML for real estate World!")
```

```python
# Step 2: Ask for user input
name = input("What is your name? ")
print(f"Nice to meet you, {name}!")
```

```python
# Step 3: Perform basic math
num1 = float(input("Enter a number: "))
num2 = float(input("Enter another number: "))
sum_result = num1 + num2
print(f"The sum of {num1} and {num2} is {sum_result}")
```

```python
# Step 4 (Upgrade): Custom function for safe input
def get_valid_number(prompt_text):
    """Asks for input until a valid number is entered."""
    while True:
        try:
            # Try to convert the input directly to a float
            number = float(input(prompt_text))
            return number
        except ValueError:
            # If it wasn't a number, we end up here
            print("Error: That was not a valid number. Please try again!\n")

def multiply_numbers(a, b):
    return a * b

print("=== Welcome to the Advanced Calculator ===")

# Main program loop
while True:
    # Step 3 (Upgrade): Using our new function for input
    num1 = get_valid_number("Enter a number: ")
    num2 = get_valid_number("Enter another number: ")

    # Calculations
    sum_result = num1 + num2
    product_result = multiply_numbers(num1, num2)

    # Output (using :.2f rounds visually to two decimal places)
    print("\n--- Your Results ---")
    print(f"The sum of {num1} and {num2} is {sum_result:.2f}")
    print(f"The product of {num1} and {num2} is {product_result:.2f}")
    print("--------------------\n")

    # Ask if the user wants to continue
    continue_calc = input("Would you like to calculate something else? (yes/no): ").lower()
    if continue_calc != "yes":
        print("Calculator shutting down. See you next time!")
        break

    # An empty line for clarity when restarting
    print("\n")
```

```python
# Step 5: Work with a list
fruits = ["apple", "banana", "cherry"]
print("\nHere's a list of fruits:", fruits)

# Add a fruit to the list
new_fruit = input("Add a fruit to the list: ")
fruits.append(new_fruit)
print("Updated list of fruits:", fruits)
```

```python
# Step 6: Loop through the list
print("\nHere's each fruit in the list:")
for fruit in fruits:
    print(fruit)
```
